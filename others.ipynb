```python
import pandas as pd
```

```python
reviews = pd.read_csv("olist_order_reviews_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
sellers = pd.read_csv("olist_sellers_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
```

```python
##repeating customers
#finding repetitive unique id
repeating_ids = customers.groupby('customer_unique_id').size()
repeating_ids = repeating_ids[repeating_ids > 1].sort_values(ascending=False)
##finding percentage of customers ordered more then one
repeatingcstm_pe = (repeating_ids.count()/customers['customer_unique_id'].nunique())*100
print(f"percentage of customers repeating  is :{repeatingcstm_pe}%")
```

```python
#sellers with worst reviews
##adding review score in order items
sellers_reviews = order_items.merge(reviews[['order_id','review_score']],on='order_id',how='left')
#grouping sellers acc to their review score

worst_reviewsellers = (
    sellers_reviews.groupby('seller_id')
    .agg(
        avg_review=('review_score', 'mean'),
        review_count=('review_score', 'count')
    )
    .sort_values('review_count', ascending=False)
)

#finding sellers with  review count>100 and their reviews
reliable_sellers = worst_reviewsellers[
    worst_reviewsellers['review_count'] >= 100
]

print(reliable_sellers.sort_values('avg_review').head(20))
```
